```python
import os
import mne
mne.set_log_level("error")
```

```python
# Set MNE configuration for light theme
mne.set_config('MNE_BROWSER_THEME', 'light')  # Set browser theme to light
mne.viz.set_browser_backend('qt')

# Alternative approach: set Qt style
os.environ['QT_STYLE_OVERRIDE'] = 'Fusion'  # Use Fusion style which is lighter
```

```python
sub = "AT36"
# dir = f"I:/Shaked/ISO_data/elderly_control_raw/{sub}/CleaningPipe/cleaned_raw_hypno_raw.fif"
path = "E:/May/May_MCI_clean/input_files/MCI003_BL_PSG_20171213_224334.mff"
```

```python
raw = mne.io.read_raw(path, preload=False)
```

```python
raw.info
```

```python
new_annot = "I:\Shaked\ISO_data\elderly_control_raw\AT36\CleaningPipe\AT36_cleaned_hypno_annotations.txt"
raw.set_annotations(mne.read_annotations(new_annot))
raw.annotations
```

```python
raw.plot()
```

```python
import yasa
spindles = yasa.spindles_detect(raw, freq_sp=(12, 15))
if spindles:
    spindles_df = spindles.summary()
    print(len(spindles_df))
    spindle_annotations = mne.Annotations(
        spindles_df['Start'].values, 
        spindles_df['Duration'].values, 
        ['Spindle_' + channel for channel in spindles_df['Channel']],
        orig_time=raw.annotations.orig_time
    )
    annotations = raw.annotations + spindle_annotations
    raw.set_annotations(annotations)
```

```python
raw.plot(duration=40)
```
